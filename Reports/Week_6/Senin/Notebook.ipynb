## Pengenalan Exploratory Data Analysis (EDA)
**Dataset:** `dataset_PenjualanRumah.xlsx` — data listing rumah dijual (harga, luas bangunan, luas tanah, kamar tidur, kamar mandi, garasi)

**Tujuan hari ini:**
- Memahami struktur dataset
- Melakukan analisis statistik dasar: mean, median, modus, minimum, maksimum, standar deviasi, kuartil
- Menggunakan `describe()`, `value_counts()`, `unique()`, `nunique()`


```python
import pandas as pd
import numpy as np

pd.set_option('display.float_format', lambda x: f'{x:,.2f}')

df = pd.read_excel('dataset_PenjualanRumah.xlsx')
print("Ukuran dataset:", df.shape)
df.head()
```

## 1. Info & Tipe Data

```python
df.info()
```

## 2. Cek Missing Value

```python
df.isna().sum()
```

## 3. Statistik Deskriptif dengan `describe()`
Kolom numerik yang dianalisis: `HARGA`, `LB` (luas bangunan), `LT` (luas tanah), `KT` (kamar tidur), `KM` (kamar mandi), `GRS` (garasi)

```python
numeric_cols = ['HARGA','LB','LT','KT','KM','GRS']
df[numeric_cols].describe()
```

## 4. Statistik Dasar Manual (Mean, Median, Modus, Min, Max, Std, Kuartil)

```python
summary = pd.DataFrame(index=numeric_cols)
summary['Mean'] = df[numeric_cols].mean()
summary['Median'] = df[numeric_cols].median()
summary['Modus'] = df[numeric_cols].mode().iloc[0]
summary['Minimum'] = df[numeric_cols].min()
summary['Maksimum'] = df[numeric_cols].max()
summary['Std Deviasi'] = df[numeric_cols].std()
summary['Q1 (25%)'] = df[numeric_cols].quantile(0.25)
summary['Q2 (50%)'] = df[numeric_cols].quantile(0.50)
summary['Q3 (75%)'] = df[numeric_cols].quantile(0.75)
summary['IQR'] = summary['Q3 (75%)'] - summary['Q1 (25%)']
summary
```

## 5. `value_counts()`, `unique()`, `nunique()`
Contoh pada kolom kategorikal/diskrit: Kamar Tidur (KT) dan Garasi (GRS)

```python
print("Jumlah kategori unik KT:", df['KT'].nunique())
print("Nilai unik KT:", sorted(df['KT'].unique()))
print()
print("Distribusi Kamar Tidur (KT):")
print(df['KT'].value_counts().sort_index())
```

```python
print("Jumlah kategori unik GRS:", df['GRS'].nunique())
print()
print("Distribusi Garasi (GRS):")
print(df['GRS'].value_counts().sort_index())
```

## 6. Insight Awal (Statistik Dasar)

- **Harga rumah** sangat bervariasi: mean ≈ Rp 7,6 miliar jauh di atas median ≈ Rp 5 miliar → distribusi harga **menceng ke kanan (right-skewed)**, dipengaruhi rumah-rumah mewah dengan harga sangat tinggi (maks Rp 65 miliar).
- **Luas bangunan (LB)** dan **luas tanah (LT)** juga right-skewed (mean > median), menandakan ada beberapa properti dengan luas jauh di atas rata-rata (LB maks 1.126 m², LT maks 1.400 m²).
- **Kamar tidur (KT)** mayoritas berkisar 4 (median = modus = 4), rentang 2–10.
- **Kamar mandi (KM)** dan **garasi (GRS)** relatif merata di sekitar 2–4 unit, dengan sebagian kecil rumah memiliki garasi hingga 10.
- IQR (Q3–Q1) pada HARGA cukup lebar (≈ Rp 5,7 miliar), mengindikasikan sebaran harga yang lebar dan berpotensi ada outlier di sisi atas — akan dianalisis lebih lanjut di **Hari Rabu (Outlier & Pattern Analysis)**.
